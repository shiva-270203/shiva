```python
# idea 
# extract key and values
# apply the for loop
# check the condition if the data type=object then
#  append in cat list
#  otherwise append in num list
```

```python
keys = visa_df.dtypes.keys()
values = visa_df.dtypes.values
cat = []
num = []
for i,j in zip(keys,values):
    print(i,j)
    if j=='object':
        cat.append(i)
    else:
        num.append(i)
```

```python
keys=visa_df.dtypes.keys()
values=visa_df.dtypes.values
cat = []
num = []
for i,j in zip(keys,values):
    print(i,j)
    if i=='object':
        cat.append(i)
    else:
        num.append(i)
```
