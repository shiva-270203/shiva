# median winsorization

```python
q1 = np.percentile(visa_df["prevailing_wage"] , 25)
q2 = np.percentile(visa_df["prevailing_wage"] , 50)
q3 = np.percentile(visa_df["prevailing_wage"] , 75)
```

# out = q3 +  1.5 * iqr

```python
iqr = q3 - q1
```

```python
upper = q3+1.5*iqr
lower = q1-1.5 * iqr
con1 = visa_df["prevailing_wage"] > upper
con2 = visa_df["prevailing_wage"] < lower
con = con1 | con2
```

```python
outliers_df = visa_df[con]
outliers_df
```

```python
outliers_values = outliers_df["prevailing_wage"].values.tolist()
median_value = np.median(outliers_values)
l = []
for i in visa_df["prevailing_wage"]:
    if i in outliers_values:
        l.append(median_value)
    else:
        l.append(i)
```
