- if we are finding something then go for **fit**

- if we are finding some new value and using those values

**One Hot Encoder**

- One hot means at a time only one value **ON**

- Other will be **OFF**

- ON means one value become **True**

- OFF means one value become **False**

- **True** means one value become **1**

- **False** means one valus become **0**


```python
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder(categories=order)
oe.fit_transform(visa_df[['education_of_employee']])
```
